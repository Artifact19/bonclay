## 2. Environment Setup

Run the cell below to install or verify dependencies. It works both in **Google Colab** (CUDA) and **Local Jupyter** (Apple Silicon MPS / CUDA / CPU).

````python
# Uncomment the pip install line below when running in a fresh Google Colab session:
# !pip install -q qwen-tts==0.1.1 soundfile librosa torchaudio

import os
import sys
import time
from pathlib import Path
import torch
import soundfile as sf
from IPython.display import Audio, display

# Attach local BonClay backend if running inside the repository
REPO_ROOT = Path.cwd().parent if Path.cwd().name == "notebook" else Path.cwd()
BACKEND_DIR = REPO_ROOT / "backend"
if BACKEND_DIR.exists() and str(BACKEND_DIR) not in sys.path:
    sys.path.insert(0, str(BACKEND_DIR))

def select_device() -> tuple[str, torch.dtype]:
    if torch.cuda.is_available():
        dtype = torch.bfloat16 if torch.cuda.is_bf16_supported() else torch.float16
        print(f"Using CUDA GPU: {torch.cuda.get_device_name(0)} ({dtype})")
        return "cuda:0", dtype
    if hasattr(torch.backends, "mps") and torch.backends.mps.is_available():
        print("Using Apple Silicon MPS (float16)")
        return "mps", torch.float16
    print("Using CPU (float32)")
    return "cpu", torch.float32

DEVICE, DTYPE = select_device()
````

## 3. Load `Qwen3-TTS-12Hz-0.6B-Base`

BonClay uses the `0.6B-Base` checkpoint by default because it fits comfortably in 16 GB unified memory alongside Kokoro and Whisper while delivering sub-second first-chunk latency.

````python
from qwen_tts import Qwen3TTSModel

# Switch to "Qwen/Qwen3-TTS-12Hz-1.7B-Base" if benchmarking on a 24GB+ GPU
MODEL_ID = os.environ.get("BONCLAY_QWEN_MODEL", "Qwen/Qwen3-TTS-12Hz-0.6B-Base")

t0 = time.perf_counter()
model = Qwen3TTSModel.from_pretrained(
    MODEL_ID,
    device_map=DEVICE,
    dtype=DTYPE,
)
print(f"Loaded {MODEL_ID} in {time.perf_counter() - t0:.2f}s")
````

## 4. Reference Audio & Reusable Voice Prompt

Instead of re-encoding the reference `.wav` on every sentence, BonClay pre-computes a **Voice Clone Prompt** (`create_voice_clone_prompt`) and caches it on disk (`voice.pt`). This cuts synthesis latency by ~40% on subsequent requests.

````python
# Point to any 10-30s clean WAV file (defaults to BonClay's bundled demo fixture when run locally)
DEFAULT_SAMPLE = REPO_ROOT / "docs" / "samples" / "en_kokoro.wav"
REF_AUDIO_PATH = str(DEFAULT_SAMPLE if DEFAULT_SAMPLE.exists() else "reference.wav")
REF_TEXT = "Hello, welcome to BonClay. Your voice, in twenty-two Indian languages."

if Path(REF_AUDIO_PATH).exists():
    print(f"Reference clip ready: {REF_AUDIO_PATH}")
    display(Audio(REF_AUDIO_PATH))
else:
    print(f"Place a 15-30s WAV file at {REF_AUDIO_PATH} and update REF_TEXT above.")
````

````python
# 1. Build a reusable speaker prompt once
voice_prompt = model.create_voice_clone_prompt(
    ref_audio=REF_AUDIO_PATH,
    ref_text=REF_TEXT,
    x_vector_only_mode=False,
)

# 2. Synthesize multiple sentences without re-encoding the reference audio
SENTENCES = [
    ("English", "Good morning! Your appointment at the clinic is confirmed for 4 PM today."),
    ("Auto", "Namaste! Aapka appointment aaj shaam chaar baje confirm ho gaya hai."),
]

for lang, text in SENTENCES:
    t0 = time.perf_counter()
    wavs, sr = model.generate_voice_clone(
        text=text,
        language=lang,
        voice_clone_prompt=voice_prompt,
    )
    elapsed = time.perf_counter() - t0
    duration = len(wavs[0]) / sr
    print(f"[{lang}] Generated {duration:.2f}s audio in {elapsed:.2f}s (RTF: {elapsed / max(duration, 1e-3):.2f}x)")
    display(Audio(wavs[0], rate=sr))
````

## 5. Extending BonClay: Adding a New Voice Engine

When you want to add a new TTS or voice-cloning model to BonClay in the future, follow this 3-step contract:

1. **Implement a Service Wrapper** following `BaseVoiceEngine` below in `backend/services/<your_engine>_service.py`.
2. **Register the Engine ID** in `backend/services/model_manager.py` (`ENGINE_REGISTRY`) so BonClay's memory guard and sidebar resource monitor can track and unload it.
3. **Add Routing Logic** in `backend/routers/tts.py` so `/api/tts/generate` and `/api/tts/stream` can dispatch to your engine.

````python
from abc import ABC, abstractmethod
from dataclasses import dataclass
from typing import Iterator, Optional
import numpy as np

@dataclass
class SynthesisResult:
    audio: np.ndarray          # 1-D float32 waveform in [-1.0, 1.0]
    sample_rate: int           # e.g. 24000
    engine_id: str             # e.g. "custom_v1"
    language: str              # ISO code e.g. "en", "hi"
    first_chunk_ms: float = 0.0

class BaseVoiceEngine(ABC):
    """Standard contract for any TTS / Voice Cloning engine in BonClay."""

    engine_id: str = "base"
    estimated_vram_mb: int = 1000
    supported_languages: tuple[str, ...] = ("en",)

    @abstractmethod
    def load(self, device: str = "cpu") -> None:
        """Load weights into memory on first use."""

    @abstractmethod
    def unload(self) -> None:
        """Release weights and clear CUDA/MPS cache when idle."""

    @abstractmethod
    def synthesize(
        self,
        text: str,
        language: str = "en",
        ref_audio_path: Optional[str] = None,
        ref_text: Optional[str] = None,
        speed: float = 1.0,
    ) -> SynthesisResult:
        """Generate full waveform synchronously."""

    def stream(
        self,
        text: str,
        language: str = "en",
        ref_audio_path: Optional[str] = None,
        ref_text: Optional[str] = None,
        speed: float = 1.0,
    ) -> Iterator[tuple[np.ndarray, int]]:
        """Yield (pcm_chunk, sample_rate) for low-latency streaming."""
        res = self.synthesize(text, language, ref_audio_path, ref_text, speed)
        yield res.audio, res.sample_rate


# Example skeleton for a new engine:
class CustomEngineExample(BaseVoiceEngine):
    engine_id = "custom_v1"
    estimated_vram_mb = 1200
    supported_languages = ("en", "hi", "ta", "te")

    def __init__(self):
        self._loaded = False

    def load(self, device: str = "cpu") -> None:
        self._loaded = True

    def unload(self) -> None:
        self._loaded = False
        if torch.cuda.is_available():
            torch.cuda.empty_cache()

    def synthesize(
        self,
        text: str,
        language: str = "en",
        ref_audio_path: Optional[str] = None,
        ref_text: Optional[str] = None,
        speed: float = 1.0,
    ) -> SynthesisResult:
        if not self._loaded:
            self.load(DEVICE)
        # Replace with your model's forward pass:
        sr = 24000
        t = np.linspace(0, 1.0, sr, endpoint=False, dtype=np.float32)
        dummy_wave = 0.1 * np.sin(2 * np.pi * 220.0 * t)
        return SynthesisResult(audio=dummy_wave, sample_rate=sr, engine_id=self.engine_id, language=language)

engine = CustomEngineExample()
res = engine.synthesize("Testing BonClay engine contract", language="en")
print(f"Engine '{res.engine_id}' produced {len(res.audio)} samples at {res.sample_rate} Hz")
````
